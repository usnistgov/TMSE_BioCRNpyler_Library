# Evaluating of txt2biocrnpyler Generated Model

This notebook executes and validates the BioCRNpyler Python script generated and downloaded directly from the `txt2biocrnpyler` tool. 

We used [`jung2022_and_gate_rxns.txt`](./jung2022_and_gate_rxns.txt) in `txt2biocrnpyler` to generate a Chemical Reaction Network (CRN) representing the DNA strand displacement **AND gate** detailed in *Programming Cell-free Biosensors with DNA Strand Displacement Circuits* ([Jung et al., Nature Chemical Biology 2022](https://www.nature.com/articles/s41589-021-00962-9)).

To verify the accuracy and functionality of the tool's output, the downloaded `.py` script is executed within this notebook without manual debugging. The resulting model is then simulated to confirm that the generated CRN correctly reproduces the expected biological AND gate logic.

```python
from pathlib import Path
# Get base directory for BioCRNpyler scripts
directory = Path.cwd()
for p in [directory, *directory.parents]:
    if p.name == 'TMSE_BioCRNpyler_Library':
        base_dir = p
        break

# Directory for models
model_dir = (base_dir / 'models').as_posix()
```

```python
"""BioCRNpyler model auto-generated by txt2biocrnpyler from 'model'.

Run this file to rebuild the ChemicalReactionNetwork and write its SBML:
    python this_script.py
"""
from biocrnpyler import (
    ChemicalReactionNetwork,
    GeneralPropensity,
    ParameterEntry,
    Reaction,
    Species,
)

# --- Species ---------------------------------------------------------
species = {
    'TetR': Species('TetR'),
    'D_tetO': Species('D_tetO'),
    'TetRD': Species('TetRD'),
    'tet': Species('tet'),
    'TetR_tet': Species('TetR_tet'),
    'RNAP': Species('RNAP'),
    'RD_tetO': Species('RD_tetO'),
    'InvR_T': Species('InvR_T'),
    'SmtB': Species('SmtB'),
    'D_smtO': Species('D_smtO'),
    'SmtBD': Species('SmtBD'),
    'Zn': Species('Zn'),
    'SmtB_Zn': Species('SmtB_Zn'),
    'RD_smtO': Species('RD_smtO'),
    'InvR_S': Species('InvR_S'),
    'AND_Gate': Species('AND_Gate'),
    'InvR_D': Species('InvR_D'),
    'Gated_TS': Species('Gated_TS'),
    'RQ': Species('RQ'),
    'SD': Species('SD'),
    'Q': Species('Q'),
}

# --- Parameters ------------------------------------------------------
parameters = {
    'k_m_tetO': ParameterEntry('k_m_tetO', 0.1),
    'k_m_smtO': ParameterEntry('k_m_smtO', 0.04),
    'k_bind': ParameterEntry('k_bind', 56.0),
    'k_unbind': ParameterEntry('k_unbind', 0.2),
    'k_SD': ParameterEntry('k_SD', 0.04),
    'k_SD_AND': ParameterEntry('k_SD_AND', 0.08),
    'k_rep_T': ParameterEntry('k_rep_T', 2.98),
    'k_derep_T': ParameterEntry('k_derep_T', 0.001),
    'k_ind_T': ParameterEntry('k_ind_T', 4.0),
    'k_unin_T': ParameterEntry('k_unin_T', 0.022),
    'k_rep_S': ParameterEntry('k_rep_S', 3.0),
    'k_derep_S': ParameterEntry('k_derep_S', 0.18),
    'k_ind_S': ParameterEntry('k_ind_S', 80.0),
    'k_unin_S': ParameterEntry('k_unin_S', 0.1),
}

# --- Reactions -------------------------------------------------------
reactions = [
    Reaction(
        inputs=[species['TetR'], species['D_tetO']],
        outputs=[species['TetRD']],
        propensity_type=GeneralPropensity(
            propensity_function='k_rep_T * TetR * D_tetO - k_derep_T * TetRD',
            propensity_species=[species['D_tetO'], species['TetR'], species['TetRD']],
            propensity_parameters=[parameters['k_derep_T'], parameters['k_rep_T']],
        ),
    ),
    Reaction(
        inputs=[species['TetR'], species['tet']],
        outputs=[species['TetR_tet']],
        propensity_type=GeneralPropensity(
            propensity_function='k_ind_T * TetR * tet - k_unin_T * TetR_tet',
            propensity_species=[species['TetR'], species['TetR_tet'], species['tet']],
            propensity_parameters=[parameters['k_ind_T'], parameters['k_unin_T']],
        ),
    ),
    Reaction(
        inputs=[species['RNAP'], species['D_tetO']],
        outputs=[species['RD_tetO']],
        propensity_type=GeneralPropensity(
            propensity_function='k_bind * RNAP * D_tetO - k_unbind * RD_tetO',
            propensity_species=[species['D_tetO'], species['RD_tetO'], species['RNAP']],
            propensity_parameters=[parameters['k_bind'], parameters['k_unbind']],
        ),
    ),
    Reaction(
        inputs=[species['RD_tetO']],
        outputs=[species['InvR_T'], species['RNAP'], species['D_tetO']],
        propensity_type=GeneralPropensity(
            propensity_function='k_m_tetO * RD_tetO',
            propensity_species=[species['RD_tetO']],
            propensity_parameters=[parameters['k_m_tetO']],
        ),
    ),
    Reaction(
        inputs=[species['SmtB'], species['D_smtO']],
        outputs=[species['SmtBD']],
        propensity_type=GeneralPropensity(
            propensity_function='k_rep_S * SmtB * D_smtO - k_derep_S * SmtBD',
            propensity_species=[species['D_smtO'], species['SmtB'], species['SmtBD']],
            propensity_parameters=[parameters['k_derep_S'], parameters['k_rep_S']],
        ),
    ),
    Reaction(
        inputs=[species['SmtB'], species['Zn']],
        outputs=[species['SmtB_Zn']],
        propensity_type=GeneralPropensity(
            propensity_function='k_ind_S * SmtB * Zn - k_unin_S * SmtB_Zn',
            propensity_species=[species['SmtB'], species['SmtB_Zn'], species['Zn']],
            propensity_parameters=[parameters['k_ind_S'], parameters['k_unin_S']],
        ),
    ),
    Reaction(
        inputs=[species['RNAP'], species['D_smtO']],
        outputs=[species['RD_smtO']],
        propensity_type=GeneralPropensity(
            propensity_function='k_bind * RNAP * D_smtO - k_unbind * RD_smtO',
            propensity_species=[species['D_smtO'], species['RD_smtO'], species['RNAP']],
            propensity_parameters=[parameters['k_bind'], parameters['k_unbind']],
        ),
    ),
    Reaction(
        inputs=[species['RD_smtO']],
        outputs=[species['InvR_S'], species['RNAP'], species['D_smtO']],
        propensity_type=GeneralPropensity(
            propensity_function='k_m_smtO * RD_smtO',
            propensity_species=[species['RD_smtO']],
            propensity_parameters=[parameters['k_m_smtO']],
        ),
    ),
    Reaction(
        inputs=[species['InvR_T'], species['InvR_S'], species['AND_Gate']],
        outputs=[species['InvR_D'], species['Gated_TS']],
        propensity_type=GeneralPropensity(
            propensity_function='k_SD_AND * InvR_T * InvR_S * AND_Gate',
            propensity_species=[species['AND_Gate'], species['InvR_S'], species['InvR_T']],
            propensity_parameters=[parameters['k_SD_AND']],
        ),
    ),
    Reaction(
        inputs=[species['InvR_D'], species['RQ']],
        outputs=[species['SD'], species['Q']],
        propensity_type=GeneralPropensity(
            propensity_function='k_SD * InvR_D * RQ',
            propensity_species=[species['InvR_D'], species['RQ']],
            propensity_parameters=[parameters['k_SD']],
        ),
    ),
]

# --- Chemical Reaction Network ---------------------------------------
crn = ChemicalReactionNetwork(
    species=list(species.values()),
    reactions=reactions,
)

if __name__ == "__main__":
    print(crn.pretty_print())
    crn.write_sbml_file(f'{model_dir}/jung2022_and_model_crn.xml')
    print("Wrote model_from_crn.xml")
```
